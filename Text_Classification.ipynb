```python
#Import libraries
import pandas as pd
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
```

```python
#Read data
df=pd.read_csv('/content/IMDB Dataset.csv.zip')
```

```python
X=df.iloc[0:100,0].values  #string review text in X
```

```python
y=df.iloc[0:100,1].values   #string review class in y
```

```python
reviews=[]  #creating empty list to store a review
```

```python
corpus=[]   #creating an empty list to store corpus containing all the reviews
```

```python
nltk.download('stopwords')
```

```python
#processing of all reviews in for loop
for ch in range(0,len(X)):
  review=re.sub('[^a-zA-Z]',' ',str(X[ch])) #removing any special character or number by replacing them by blank space
  #converting text to lowercase
  review=review.lower()
  #split sentences into words by split it over blank space
  review=review.split()
  #extract root words and removing stop words
  ps=PorterStemmer()
  review=[ps.stem(word) for word in review if not word in set(stopwords.words('english'))]
  review=' '.join(review)
  #all processed reviews are appended into a corpus
  corpus.append(review)
```

```python
#Applying Count Vectorizer for converting strings to vector
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer(max_features=1500)
X=cv.fit_transform(corpus).toarray()
```

```python
#Applying TF-IDF for converting strings to vector
# from sklearn.feature_extraction.text import TfidfVectorizer
# vectorizer=TfidfVectorizer()
# X=vectorizer.fit_transform(corpus).toarray()
```

```python
#Split dataset for training and testing
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=0)
```

```python
#Fitting Naive Bayes to training data
# from sklearn.naive_bayes import GaussianNB
# classifier=GaussianNB()
# classifier.fit(X_train,y_train)
```

```python
#Fitting random forest
from sklearn.ensemble import RandomForestClassifier
classifier=RandomForestClassifier(n_estimators=100,criterion='entropy')
classifier.fit(X_train,y_train)
```

```python
y_pred=classifier.predict(X_test)
```

```python
#Confusion Matrix
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(y_test,y_pred)
#Accuracy
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(y_test,y_pred)
print('Accuracy is ',accuracy)
#Precision
from sklearn.metrics import precision_score
precision = precision_score(y_test,y_pred, pos_label='positive')
print('Precision is ',precision)
#Recall
from sklearn.metrics import recall_score
recall = recall_score(y_test,y_pred, pos_label='positive')
print('Recall is ',recall)
```

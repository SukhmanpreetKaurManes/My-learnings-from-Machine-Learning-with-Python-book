DOG CAT CLASSIFIER

```python
#Import Libraries
from keras.models import Sequential
from keras.layers import Conv2D
from keras.layers import MaxPooling2D
from keras.layers import Flatten
from keras.layers import Dense
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
#Intializing the CNN
classifier = Sequential()
```

```python
#Convolution
classifier.add(Conv2D(filters=32,kernel_size=3,input_shape=[128,128,3],activation='relu'))
```

```python
#Pooling
classifier.add(MaxPooling2D(pool_size=2,strides=2))
```

```python
#Flatten
classifier.add(Flatten())
```

```python
#Full Connection
classifier.add(Dense(units=128,activation='relu'))
```

```python
#Output layer
classifier.add(Dense(units=1,activation='sigmoid'))
```

```python
#Compiling the CNN
classifier.compile(optimizer="adam",loss="binary_crossentropy",metrics=["accuracy"])
```

```python
import zipfile
import os


train_zip_path = '/content/train-20260717T022926Z-1-001.zip'
test_zip_path = '/content/test-20260717T022303Z-1-001.zip'

base_data_dir = '/content/dataset'
train_dir = os.path.join(base_data_dir, 'train')
validation_dir = os.path.join(base_data_dir, 'test')

os.makedirs(base_data_dir, exist_ok=True)

print(f"Extracting {train_zip_path} to {train_dir}...")
with zipfile.ZipFile(train_zip_path, 'r') as zip_ref:



    zip_ref.extractall(base_data_dir)

print(f"Extracting {test_zip_path} to {validation_dir}...")
with zipfile.ZipFile(test_zip_path, 'r') as zip_ref:
    zip_ref.extractall(base_data_dir)

print("Extraction complete. Verifying directory structure...")
print(f"Contents of {base_data_dir}: {os.listdir(base_data_dir)}")
print(f"Contents of {train_dir} (first 5 items): {os.listdir(train_dir)[:5]}")
print(f"Contents of {validation_dir} (first 5 items): {os.listdir(validation_dir)[:5]}")

# Image Augmentation
train_datagen = ImageDataGenerator(rescale = 1./255,shear_range=0.2,zoom_range=0.2,horizontal_flip=True)
```

```python
#code to create train_generator
train_generator = train_datagen.flow_from_directory(train_dir,target_size=(128,128),batch_size=32,class_mode='binary')
```

```python
# Image Augmentation for test Data
test_datagen = ImageDataGenerator(rescale = 1./255)

# Code to create validation_generator
test_generator = test_datagen.flow_from_directory(
    validation_dir,
    target_size=(128, 128),
    batch_size=32,
    class_mode="binary"
)
```

```python
#Training the CNN on the training set and evaluating it on test set
classifier.fit(train_generator,steps_per_epoch = 100, epochs = 50,validation_data=test_generator,validation_steps=800)
```

```python
import numpy as np
from keras.preprocessing import image

unseen_image = image.load_img("/content/mat.jpg",target_size=(128,128))
unseen_image = image.img_to_array(unseen_image)
unseen_image = np.expand_dims(unseen_image,axis=0)

# Normalize the image, as done during training
unseen_image = unseen_image / 255.0

result = classifier.predict(unseen_image)

# Get class labels from the training generator
class_labels = {v: k for k, v in train_generator.class_indices.items()}

# The classifier outputs a probability. For binary classification with sigmoid, if result[0][0] > 0.5, it's class 1, otherwise class 0.
# We need to map this numerical prediction back to the class label.
if result[0][0] > 0.5:
  prediction_index = 1
else:
  prediction_index = 0

prediction = class_labels[prediction_index]

print("The given image is of " + prediction)
```
